## 1. Import library

```python
import math, itertools, random, statistics
from fractions import Fraction
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import sympy as sp
import scipy
from scipy import stats
from sklearn.naive_bayes import GaussianNB

print("NumPy :", np.__version__)
print("SciPy :", scipy.__version__)
print("Pandas:", pd.__version__)
```

## 2. `math`: faktorial, permutasi, kombinasi

`math.factorial(n)` = n!, `math.perm(n, r)` = nPr, `math.comb(n, r)` = nCr.

```python
print("5!       =", math.factorial(5))
print("4P2      =", math.perm(4, 2))
print("4C2      =", math.comb(4, 2))
print("e^(-2)   =", round(math.exp(-2), 4))
```

## 3. `itertools`: membuat ruang sampel

`product` (dengan pengembalian, urutan penting), `permutations` (tanpa pengembalian, urutan penting), `combinations` (tanpa pengembalian, urutan tidak penting), `combinations_with_replacement` (dengan pengembalian, urutan tidak penting).

```python
print("product      :", list(itertools.product("HT", repeat=2)))
print("permutations :", list(itertools.permutations("ABCD", 2)))
print("combinations :", list(itertools.combinations("ABCD", 2)))
print("comb. w/ repl:", len(list(itertools.combinations_with_replacement("ABCD", 2))), "kemungkinan")
```

## 4. `fractions`: probabilitas sebagai pecahan eksak

```python
p = Fraction(2, 4)
q = Fraction(1, 4)
print("p + q       =", p + q, "=", float(p + q))
print("3/4 x 2/3   =", Fraction(3, 4) * Fraction(2, 3))
```

## 5. `random` dan `statistics`

```python
random.seed(42)
print("choice :", random.choice(["H", "T"]))
print("sample :", random.sample(range(1, 11), 3))

data = [2, 3, 8, 4, 5]
print("mean      :", statistics.mean(data))
print("variance  :", statistics.variance(data))
print("pstdev    :", round(statistics.pstdev(data), 4))
print("mode      :", statistics.mode([1, 2, 2, 3]))
```

## 6. `numpy`: simulasi dan operasi array

`np.random.default_rng(seed)` membuat generator acak yang hasilnya bisa diulang.

```python
rng = np.random.default_rng(42)
print("binomial :", rng.binomial(n=10, p=0.5, size=5))
print("normal   :", rng.normal(0, 1, 3).round(3))
print("mean, var:", np.mean([1, 2, 3, 4]), np.var([1, 2, 3, 4]))
```

## 7. `scipy.stats`: pustaka utama distribusi

Pola umum: `distribusi.pmf/pdf(x, ...)`, `.cdf(x, ...)`, `.ppf(q, ...)` (kuantil), `.rvs(size=...)` (acak), `.stats(..., moments='mv')` (mean dan varians).

```python
print("Binomial   P(X=3), n=5, p=0.25 :", round(stats.binom.pmf(3, 5, 0.25), 4))
print("Poisson    P(X=1), lambda=2    :", round(stats.poisson.pmf(1, 2), 4))
print("Normal     P(X<15), mu=12, s=3 :", round(stats.norm.cdf(15, loc=12, scale=3), 4))
print("Normal     x untuk P=0.8       :", round(stats.norm.ppf(0.8, 12, 3), 4))
mean, var = stats.binom.stats(5, 0.25, moments="mv")
print("Binomial   mean, var           :", mean, var)
```

## 8. `pandas`: tabel frekuensi dan tabel kontingensi

```python
df = pd.DataFrame({
    "os":      ["Android", "Android", "iOS", "iOS", "Android"],
    "premium": ["ya", "tidak", "ya", "ya", "tidak"],
})
print(pd.crosstab(df["os"], df["premium"]))
print()
print(pd.crosstab(df["os"], df["premium"], normalize="all"))
```

## 9. `matplotlib` dan `seaborn`: visualisasi

```python
x = np.arange(0, 6)
plt.figure(figsize=(6, 3.5))
plt.bar(x, stats.binom.pmf(x, 5, 0.25), color="steelblue")
plt.title("Contoh: Binomial(n=5, p=0.25)")
plt.xlabel("Jumlah sukses")
plt.ylabel("Probabilitas")
plt.show()
```

## 10. `sympy`: perhitungan simbolik

```python
p = sp.symbols("p")
print("C(5,3)       =", sp.binomial(5, 3))
print("3/4 + 1/4    =", sp.Rational(3, 4) + sp.Rational(1, 4))
print("Rumus binomial simbolik:", sp.binomial(5, 3) * p**3 * (1 - p)**2)
print("Nilai pada p=1/4       :", (sp.binomial(5, 3) * p**3 * (1 - p)**2).subs(p, sp.Rational(1, 4)))
```

## 11. `scikit-learn`: Naive Bayes

Contoh sangat kecil: dua kelompok data satu dimensi, lalu model memprediksi kelas data baru.

```python
X = np.array([[1.0], [1.2], [3.0], [3.2]])
y = [0, 0, 1, 1]
model = GaussianNB().fit(X, y)
print("Prediksi untuk 1.1 dan 3.1 :", model.predict([[1.1], [3.1]]))
print("Probabilitas kelas utk 2.1 :", model.predict_proba([[2.1]]).round(3))
```

# Bab 2: Probabilitas Dasar: Pendekatan, Ruang Sampel, Kejadian, Komplemen

Isi: tiga pendekatan probabilitas (klasik, empiris, subjektif), ruang sampel dan kejadian, komplemen, serta hukum bilangan besar. Aturan dasar: **0 ≤ P(A) ≤ 1**, dan **P(S) = 1**.

## Pengertian singkat

- **Ruang sampel (S):** himpunan semua hasil yang mungkin dari suatu percobaan. **Titik sampel:** masing-masing anggota S.
- **Kejadian (event):** himpunan bagian dari S.
- **Pendekatan klasik:** P(A) = n(A) / n(S), dengan syarat semua hasil sama mungkin.
- **Pendekatan empiris:** P(A) = f(A) / n, dari frekuensi relatif hasil pengamatan.
- **Pendekatan subjektif:** P(A) berasal dari penilaian pakar/ahli, bukan hitungan data.

```python
import itertools
from fractions import Fraction
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

## Soal 2.1: Pendekatan klasik

**Studi kasus:** Sebuah tim DevOps mengelola 10 server. Sebanyak 3 server sedang dalam masa maintenance. Satu server dipilih secara acak untuk di-deploy.

**Ditanyakan:** Berapa probabilitas server yang terpilih sedang maintenance?

**Konsep / rumus:** P(A) = n(A) / n(S)

**Perhitungan manual:** n(A) = 3, n(S) = 10, maka P(A) = 3/10 = 0,3.

```python
n_total = 10
n_A = 3
P = Fraction(n_A, n_total)
print("Pecahan   :", P)
print("Desimal   :", float(P))
print("Persentase:", f"{float(P):.0%}")
```

**Interpretasi:** Angka 0,3 berarti dari setiap 10 kali pemilihan acak, rata-rata ada 3 kali terpilih server yang sedang maintenance. Nilainya berada di antara 0 dan 1 sehingga valid. Hasil ini hanya layak jika setiap server memang punya peluang terpilih yang sama (asumsi pendekatan klasik).

## Soal 2.2: Ruang sampel dan kejadian: pelemparan koin dua kali

**Studi kasus:** Sebuah koin dilempar dua kali (T = tail, H = head).

**Ditanyakan:** Tentukan ruang sampel dan probabilitas kedua lemparan menghasilkan sisi yang sama.

**Konsep / rumus:** S = semua pasangan hasil; A = kejadian dua sisi sama.

**Perhitungan manual:** S = {(T,T), (T,H), (H,T), (H,H)}, n(S) = 4. A = {(T,T), (H,H)}, n(A) = 2, maka P(A) = 2/4 = 1/2 = 50%.

```python
S = list(itertools.product("TH", repeat=2))
print("Ruang sampel :", S, "| n(S) =", len(S))

A = [e for e in S if e[0] == e[1]]
print("Kejadian A   :", A, "| n(A) =", len(A))
print("P(A)         :", Fraction(len(A), len(S)), "=", len(A) / len(S))
```

**Interpretasi:** Ruang sampel berisi 4 titik sampel, dan 2 di antaranya memenuhi kejadian A, sehingga P(A) = 0,5. Artinya peluang dua sisi sama sebanding dengan peluang dua sisi berbeda. Hasil kode (cacah dengan `itertools.product`) sama dengan hitungan manual.

## Soal 2.3: Ruang sampel tiga server (y/n)

**Studi kasus:** Tiga server (Sta-1, Sta-2, Sta-3) dicek apakah perlu patch keamanan (y = perlu, n = tidak perlu). Satu hasil pengecekan ditulis sebagai (y,n,y), artinya server 1 dan 3 perlu patch.

**Ditanyakan:** Tentukan ruang sampel, lalu probabilitas kejadian A (paling sedikit 2 server perlu patch), B (tidak ada server yang perlu patch), dan C (tepat 2 server perlu patch).

**Konsep / rumus:** Pendekatan klasik dengan asumsi 8 hasil sama mungkin.

**Perhitungan manual:** n(S) = 2³ = 8. A = 4 elemen → 4/8 = 0,5. B = {(n,n,n)} → 1/8 = 0,125. C = 3 elemen → 3/8 = 0,375.

```python
S = list(itertools.product("yn", repeat=3))
print("n(S) =", len(S))
print(S)

A = [e for e in S if e.count("y") >= 2]
B = [e for e in S if e.count("y") == 0]
C = [e for e in S if e.count("y") == 2]

for nama, E in [("A: paling sedikit 2 perlu patch", A),
                ("B: tidak ada yang perlu patch  ", B),
                ("C: tepat 2 perlu patch         ", C)]:
    print(nama, "| n =", len(E), "| P =", Fraction(len(E), len(S)), "=", round(len(E) / len(S), 4))
```

**Interpretasi:** Peluang paling sedikit 2 server perlu patch adalah 0,5 (50%), sedangkan peluang tidak ada satu pun yang perlu patch hanya 0,125 (12,5%). Dari sisi operasional, artinya tim sebaiknya menyiapkan jadwal patch karena kasus 'tidak ada yang perlu' tergolong jarang. Catatan: asumsi 8 hasil sama mungkin hanya wajar jika kebutuhan patch tiap server tidak saling mempengaruhi dan peluangnya sama; jika tidak, pendekatan empiris lebih cocok.

## Soal 2.4: Pendekatan empiris: tabel frekuensi dan histogram

**Studi kasus:** Selama 30 jadwal, keterlambatan job batch malam (menit) dicatat sebagai berikut. Contoh ini mengikuti pola data keterlambatan bus pada modul dosen, dengan konteks job batch.

**Ditanyakan:** Hitung frekuensi relatif tiap keterlambatan, P(keterlambatan ≤ 3 menit), P(keterlambatan > 5 menit), dan rata-rata keterlambatan.

**Konsep / rumus:** P(A) = f(A) / n

**Perhitungan manual:** P(≤3) = (2+3+8)/30 = 13/30 ≈ 0,4333. P(>5) = (3+2+0+1+2)/30 = 8/30 ≈ 0,2667. Rata-rata = 134/30 ≈ 4,47 menit.

```python
delay = list(range(1, 11))
freq  = [2, 3, 8, 4, 5, 3, 2, 0, 1, 2]

df = pd.DataFrame({"keterlambatan_menit": delay, "frekuensi": freq})
n = df["frekuensi"].sum()
df["frek_relatif"] = (df["frekuensi"] / n).round(4)
print(df)
print("Total pengamatan:", n, "| jumlah frek. relatif:", df["frek_relatif"].sum().round(2))

p_le3 = df.loc[df["keterlambatan_menit"] <= 3, "frekuensi"].sum() / n
p_gt5 = df.loc[df["keterlambatan_menit"] > 5, "frekuensi"].sum() / n
rata2 = (df["keterlambatan_menit"] * df["frekuensi"]).sum() / n
modus = df.loc[df["frekuensi"].idxmax(), "keterlambatan_menit"]
print(f"P(<=3 menit) = {p_le3:.4f}")
print(f"P(>5 menit)  = {p_gt5:.4f}")
print(f"Rata-rata    = {rata2:.2f} menit | Modus = {modus} menit")

plt.figure(figsize=(7, 3.5))
plt.bar(df["keterlambatan_menit"], df["frek_relatif"], color="darkorange")
plt.xticks(delay)
plt.xlabel("Keterlambatan (menit)")
plt.ylabel("Frekuensi relatif")
plt.title("Histogram keterlambatan job batch (30 jadwal)")
plt.show()
```

**Interpretasi:** Angka 0,4333 artinya sekitar 43% jadwal terlambat tidak lebih dari 3 menit, sedangkan sekitar 27% terlambat lebih dari 5 menit. Modus 3 menit menunjukkan keterlambatan paling sering. Data ini cukup layak sebagai perkiraan awal, tetapi hanya 30 pengamatan sehingga probabilitasnya masih berupa perkiraan; sampel lain bisa memberi angka yang sedikit berbeda (inilah sebabnya definisi empiris memakai limit n → ∞). Distribusinya tidak mulus (ada keterlambatan 8 menit dengan frekuensi 0), tanda sampel belum besar.

## Soal 2.5: Pendekatan subjektif

**Studi kasus:** Dua senior engineer diminta menaksir peluang rilis aplikasi tepat waktu. Engineer 1 menaksir 0,7 dan engineer 2 menaksir 0,9. Belum ada data historis rilis.

**Ditanyakan:** Berapa taksiran peluang rilis tepat waktu (rata-rata pakar) dan peluang terlambat?

**Konsep / rumus:** P(A) ditentukan dari penilaian pakar; peluang komplemen P(Aᶜ) = 1 − P(A).

**Perhitungan manual:** Rata-rata = (0,7 + 0,9)/2 = 0,8. Peluang terlambat = 1 − 0,8 = 0,2.

```python
penilaian = [0.7, 0.9]
p_tepat = sum(penilaian) / len(penilaian)
print("P(rilis tepat waktu) =", round(p_tepat, 2))
print("P(rilis terlambat)   =", round(1 - p_tepat, 2))
```

**Interpretasi:** Angka 0,8 adalah tingkat keyakinan para pakar, bukan hasil perhitungan dari data. Nilainya valid (0 sampai 1), tetapi bersifat subjektif sehingga perlu dikalibrasi dengan data rilis sebelumnya begitu tersedia. Perbedaan taksiran antar pakar (0,7 vs 0,9) menunjukkan adanya ketidakpastian.

## Soal 2.6: Komplemen kejadian

**Studi kasus:** Sebuah layanan memakai 5 server yang bekerja independen. Setiap server berpeluang 2% mengalami down dalam sehari.

**Ditanyakan:** Berapa probabilitas minimal satu server down dalam sehari?

**Konsep / rumus:** P(A) = 1 − P(Aᶜ). Di sini Aᶜ = tidak ada server yang down.

**Perhitungan manual:** P(Aᶜ) = 0,98⁵ ≈ 0,9039, sehingga P(A) = 1 − 0,9039 = 0,0961.

```python
p_down = 0.02
n_server = 5
p_tak_ada_down = (1 - p_down) ** n_server
p_minimal_satu = 1 - p_tak_ada_down
print("P(tidak ada yang down)   =", round(p_tak_ada_down, 4))
print("P(minimal satu down)     =", round(p_minimal_satu, 4))
```

**Interpretasi:** Meskipun satu server hanya 2% berpeluang down, peluang minimal satu dari lima server down sudah sekitar 9,6% per hari, hampir 5 kali lebih besar. Pelajaran untuk sistem terdistribusi: makin banyak komponen, makin besar peluang ada yang bermasalah, sehingga perlu redundansi dan monitoring. Menghitung lewat komplemen jauh lebih mudah daripada menjumlahkan banyak kasus.

## Soal 2.7: Hukum bilangan besar (definisi frekuensi relatif sebagai limit)

**Studi kasus:** Pelemparan koin adil disimulasikan dengan jumlah lemparan n = 10, 100, 1.000, 10.000, dan 100.000.

**Ditanyakan:** Bagaimana frekuensi relatif sisi head berubah ketika n bertambah?

**Konsep / rumus:** P(A) = lim (n → ∞) f(A)/n

**Perhitungan manual:** Nilai teoretis P(head) = 0,5.

```python
rng = np.random.default_rng(42)

print("   n    | frekuensi relatif head | selisih terhadap 0.5")
for n in [10, 100, 1000, 10000, 100000]:
    lemparan = rng.integers(0, 2, size=n)      # 1 = head
    fr = lemparan.mean()
    print(f"{n:>7} | {fr:>22.4f} | {abs(fr - 0.5):.4f}")

lemparan = rng.integers(0, 2, size=5000)
kumulatif = np.cumsum(lemparan) / np.arange(1, 5001)
plt.figure(figsize=(7, 3.5))
plt.plot(kumulatif, color="teal")
plt.axhline(0.5, color="red", linestyle="--", label="P teoretis = 0,5")
plt.xlabel("Jumlah lemparan (n)")
plt.ylabel("Frekuensi relatif head")
plt.title("Konvergensi frekuensi relatif")
plt.legend()
plt.show()
```

**Interpretasi:** Pada n kecil frekuensi relatif bisa jauh dari 0,5, tetapi secara umum makin besar n kurva makin mendekati garis 0,5 (tidak selalu turun mulus: selisih pada n = 10.000 sedikit lebih besar daripada n = 1.000). Ini menjelaskan mengapa estimasi empiris dari sampel kecil (seperti 30 jadwal pada Soal 2.4) kurang andal. Dalam praktik TI, simulasi dengan banyak iterasi dipakai untuk mendekati probabilitas yang sulit dihitung manual. Hasil acak ditetapkan dengan `seed` agar dapat diulang.
